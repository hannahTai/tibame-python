```python
# 雙層迴圈: 建議由內到外思考
# 推推 for i in range(n)

# 1
# hello
# hello
# 2
# hello
# hello
# 3
# hello
# hello

for i in range(3):
    print(i+1)
    for j in range(2):
        print('hello')
```

```python
def print2line():
    for j in range(2):
        print('hello')

for i in range(3):
    print(i+1)
    print2line()
```

```python
# 改成while(My)
i = 0
while i < 3:
    print(i+1)
    j = 0
    while j < 2:
        print('hello')
        j+=1
    i+=1
```

```python
# XXOOXOX(My)
# xoxox
# oxoxo
# xoxox
# oxoxo

for i in range(4):
    x = i % 2
    str = ''
    for j in range(5):
        a = (j + x) % 2
        if a == 1:
            str += "o"
        else:
            str += "x"
    print(str)
```

```python
# XXOOXOX(Teacher)
# xoxox
# oxoxo
# xoxox
# oxoxo

row, col = 4, 6
for i in range(row):
    line = ''
    for j in range(col):
        pos = j + i
        if pos % 2 == 0:
            line += "x"
        else:
            line += "o"
    print(line)
```

```python
# abcdloop(Teacher)
# abcdab
# bcdabc
# cdabcd
# dabcda

symbols = ["a", "b", "c", "d"]
row, col = 4, 6
for i in range(row):
    line = ''
    # i
    # 0 j=0,1,2; i+j=0,1,2
    # 1 j=0,1,2; i+j=1,2,3
    for j in range(col):
        pos = (i + j) % len(symbols)
        line += symbols[pos]
    print(line)
```

```python
# 單層list初始化

# 1
x = [0, 1] * 3
print(x) # [0, 1, 0, 1, 0, 1]

# 2
x = []
for i in range(5):
    n = i + 1
    x.append(n)
print(x) # [1, 2, 3, 4, 5]

# 3: 2的簡化版
x = [(i + 1) for i in range(5)]
print(x) # [1, 2, 3, 4, 5]
```
