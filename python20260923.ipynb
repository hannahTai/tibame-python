```python
import urllib.request as req

url = "https://raw.githubusercontent.com/Elwing-Chou/tibame20260826/refs/heads/main/total.csv"
req.urlretrieve(url, "total.csv")
```

```python
import pandas as pd
df = pd.read_csv("total.csv", encoding="utf-8")
df
```

```python
fil = [True, False, True]
df.head(3)[fil]
```

```python
# unique()/value_counts()
df["userid"].value_counts()
# userid
# asdf89aglkj	343
# trista771116	308

df["userid"].unique()
# array(['k9219059', 'Amy20080403', 'winolee', ..., 'huanqaq708', 'fire918',
#        'wasd123121'], dtype=object)
```

```python
import os

# 建立資料夾
dn = "analyse"
if not os.path.exists(dn):
        os.makedirs(dn)

# 取出不重複的userid
uids = df["userid"].unique()

# loop userid
for uid in uids:
    # 產生過濾器
    fil = df["userid"] == uid # True,False...準備跟資料筆數一樣多的True/False
    # 過濾完的資料，存成csv
    df[fil].to_csv(f"{dn}/{uid}.csv", encoding="utf-8", index=False)
```

- pandas 兩大操作
    - 過濾: 移除不要的資料
        - 準備跟資料筆數一樣多的True/False, []帶入DataFrame

    - 轉換: 把整行轉換成另外一種格式
