```python
import urllib.request as req
import json

# url = "https://api.gamer.com.tw/anime/v1/danmu.php?videoSn=51010&geo=TW%2CHK";
url = "https://api.gamer.com.tw/anime/v1/danmu.php?videoSn=26850&geo=TW%2CHK";

f = req.urlopen(url)
content = f.read()

print(content)
# b'{"data":{"danmu":[{"text":"\\u770b\\u5b8c2\\u52372\\/15"
# b => bytes型態 0101000
print(type(content))
# <class 'bytes'>

content = json.loads(content)

print(content)
# {'data': {'danmu': [{'text': '看完2刷2/15',
print(type(content))
# <class 'dict'>

print(content["data"])
print(content["data"]["danmu"])

danmus = content["data"]["danmu"]

for d in danmus:
    uid = d["userid"]
    text = d["text"]
    print(uid, ':', text)

f.close()
```
