````python
import urllib.request as req
import json
import pandas as pd
import os

def download_danmu(sn):

    # 建立資料夾
    dn = "baha"
    if not os.path.exists(dn):
        os.makedirs(dn)

    url = f"https://api.gamer.com.tw/anime/v1/danmu.php?videoSn={sn}&geo=TW%2CHK";

    f = req.urlopen(url)
    content = f.read()
    content = json.loads(content)
    f.close()

    danmus = content["data"]["danmu"]

    df = pd.DataFrame(danmus)

    # 擴充欄位/增加一欄
    df["sn"] = sn

    df.to_csv(f"{dn}/{sn}.csv", encoding="utf-8", index=False)
    # index=False: 不存列編號
    # df.to_excel("baha.xlsx")

    print(sn)

download_danmu(51010) #無職轉生
download_danmu(26850) #鬼滅之刃

````

- HTML格式: 網頁格式
    - 定義區塊
        - 名字: 決定區塊是什麼
        - a: 超連結
        - img: 圖片
        - video: 影片
    - 屬性: 決定細節
        - a href="網址" (go to a new page; hyberlink)
        - img/video src="網址" (show on the page: source)
    - 顯示文字
        - 區塊中間
    - sampe
        ```
        <a href="網址">顯示文字</a>
        ```



- HTTP status code(狀態碼)
    - 2開頭: 成功
    - 3開頭: 轉址
    - 4開頭: 本地端錯誤
        - 404: Not Found (網址打錯)
        - 403: Forbidden (IP被拒絕)
            - IP ban: 送得太頻繁, 被ban一段時間 => 修正程式,減緩頻率
            - Missing Header: 第一次就被拒絕 => 補Headers,常檢查user-agent
    - 5開頭:

````python
import urllib.request as req
# 第三方: BeautifulSoup4 (美麗湯)
import bs4 as bs

url = "https://ani.gamer.com.tw/animeVideo.php?sn=51010";
# 繞道練習
url = "https://github.com/Elwing-Chou/tibame20260826/raw/refs/heads/main/baha0912.html";
h = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"
}

r = req.Request(url, headers=h)
f = req.urlopen(url)
content = f.read()
f.close()

# beatufulSoup
# find: 找一個; find_all: 找全部
# 萃取,找到需要的區塊
## 屬性: [屬性名稱] -> ["href"]
## 內容: .get_text()
html = bs.BeautifulSoup(content)
for link in html.find_all("a"):
    link_href = link["href"]
    if link_href.startswith("?sn"):
        sn = link_href.split("=")[-1]
        download_danmu(sn)

````
