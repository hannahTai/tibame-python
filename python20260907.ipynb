- python四大群集
    - list: 清單(同類型) [20,30,40]
        - 有序: key=0,1,2,3
    - dist: 字典(複雜單位) {"name":"Han", "age":25}
        - 無序: key自訂=name, age
    - set: 不重複的集合 {20,30,40}
        - 無序: key無法自訂
    - tuple

```python
# Set

s1 = {20,30,40,40}
s2 = {10,20}
print(s1)
print(s2)

# 交集(&): 列出共有的
# 聯集(|): 列出所有
# 減集(-): 把有的減掉(順序會有差)
print(s1 & s2) # {20}
print(s1 | s2) # {20, 40, 10, 30}
print(s1 - s2) # {40, 30}
print(s2 - s1) # {10}

# 是否存在
print(10 in s2)

# Loop
for n in (s1 | s2):
    print("!", n)

# set add element
s1 = {20,30,40}
s1.add(99)
s1.add(40)
print(s1)

# 創建空的set
s1 = {} #dict 直接大括號會是字典
print(type(s1))
s1 = set()
print(type(s1))

# 怕不好記的話,可以都用建構子
a = dict()
print(type(a))
a = dict()
print(type(a))
a = list()
print(type(a))
```

```python
# 產生6個不重複的號碼+一個特別號 list (Teacher)
import random
lottery = []
while len(lottery) < 6:
    n = random.randint(1,49)
    print(n)
    if not n in lottery:
        lottery.append(n)
print(lottery)
```

```python
# 產生6個不重複的號碼+一個特別號 set (My)
import random
lottery = set()
while len(lottery) < 6:
    n = random.randint(1,49)
    print(n)
    lottery.add(n)
print(lottery)
```

```python
# 產生6個不重複的號碼+一個特別號 set (Teacher)
import random
lottery = set()

# 一般號
while len(lottery) < 6:
    n = random.randint(1,49)
    lottery.add(n)

# 特別號
special = None
while True:
    special = random.randint(1, 49)
    if not special in lottery:
        break

print(lottery)
print(special)
```
