# LAB 1. 完全平方日

- 完全平方日: 日期 yyyymmdd 是某數的平方
- 題目: 給你一個日期，你要往後數出接下來的10個完全平方日的日期
- 以後只要看到有^2 ^3 ^4..., 都應該使用原數字處理
- 閏年概念:
$365+\frac{1}{4}-\frac{1}{100}+\frac{1}{400} - ...$

```python
# 閏年判斷1
# 1963: X
# 1964: O
# 1800: X
# 2000: O

year = 2000
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, ": O")
        else:
            print(year, ": X")
    else:
        print(year, ": O")
else:
    print(year, ": X")
```

```python
# 閏年判斷2(suggested)
# 1963: X
# 1964: O
# 1800: X
# 2000: O

year = 2000
if year % 400 == 0:
    print(year, ": O")
elif year % 100 == 0:
    print(year, ": X")
elif year % 4 == 0:
    print(year, ": O")
else:
    print(year, ": X")
```

```python
# 閏年判斷3
# 從某年找十個閏年
start = 1992
count = 0
i = 0
while True:
    year = start + i
    if year % 4 == 0 and ((year % 100 != 0) or (year % 100 == 0 and year % 400 == 0)):
        print(year, ": 0")
        count +=1
    if(count >=10):
        break;
    i += 1
```
