````python
import urllib.request as req
import bs4 as bs

def get_page(page_num):
    url = f"https://tabelog.com/tw/tokyo/rstLst/RC0123/{page_num}?SrtT=rt"
    print(url)
    resp = req.urlopen(url)
    content = resp.read()
    html = bs.BeautifulSoup(content)

    # html有沒有抓到完成的source
    ## 有 ok
    ## 沒有 加header

    # 建議抓完後續再處理

    rs = html.find_all("div", {"class":"list-rst__body"}) # 區塊

    table = []
    for r in rs:
        name = r.find("a", {"class": "list-rst__rst-name-target"}) # 超連結
        area_genre = r.find("div", {"class": "list-rst__area-genre"}) # 地點/種類
        rating = r.find("span", {"class": "c-rating__val"}) # 評價
        prices = r.find_all("span", {"class": "c-rating-v3__val"}) # 晚間價格+白天價格
        holiday = r.find("span", {"class": "list-rst__holiday-text"}) # 休假日
        name_text = name.get_text()
        name_href = name["href"]
        area_genre_text = area_genre.get_text().strip()
        rating_text = rating.get_text().strip()
        dinner_price_text = prices[0].get_text().strip()
        lunch_price_text = prices[1].get_text().strip()
        holiday_text = holiday.get_text().strip()

        row = {
            "rating":rating_text,
            "name":name_text,
            "url":name_href,
            "area_genre":area_genre_text,
            "lunch_price":lunch_price_text,
            "dinner_price":dinner_price_text,
            "holiday":holiday_text,
        }
        table.append(row)

    return table
````

````python
import pandas as pd

total = []
for i in range(10):
    page = i + 1
    partial_df = pd.DataFrame(get_page(page))
    total.append(partial_df)
df = pd.concat(total)
df
````

````python
def func(s):
    s_spl = s.split(" / ")
    ans = {
        "area": "-",
        "genre": "-"
    }
    if len(s_spl) == 2:
        ans["area"] = s_spl[0]
        ans["genre"] = s_spl[1]
    elif len(s_spl) == 1:
        ans["genre"] = s_spl[0]
    return pd.Series(ans)

df[["area", "genre"]] = df["area_genre"].apply(func)
df
````
