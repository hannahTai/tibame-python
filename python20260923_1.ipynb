````python
import urllib.request as req
import bs4 as bs

url = "https://tabelog.com/tw/tokyo/rstLst/RC0123/?SrtT=rt"
resp = req.urlopen(url)
content = resp.read()
html = bs.BeautifulSoup(content)
# print(html)

# html有沒有抓到完成的source
## 有 ok
## 沒有 加header

# 建議抓完後續再處理

rs = html.find_all("div", {"class":"list-rst__body"}) # 區塊

for r in rs:
    name = r.find("a", {"class": "list-rst__rst-name-target"}) # 超連結
    area_genre = r.find("div", {"class": "list-rst__area-genre"}) # 地點/種類
    rating = r.find("span", {"class": "c-rating__val"}) # 評價
    price = r.find_all("span", {"class": "c-rating-v3__val"}) # 評價

    holiday = r.find("span", {"class": "list-rst__holiday-text"}) # 休假日
    name_text = name.get_text()
    name_href = name["href"]
    area_genre_text = area_genre.get_text().strip()
    rating_text = rating.get_text()
    night_price = price[0].get_text()
    lunch_price = price[1].get_text()
    holiday_text = holiday.get_text()

    print(name_text)
    print(name_href)
    print(area_genre_text)
    print(rating_text)
    print(night_price)
    print(lunch_price)
    print(holiday_text)
    print("-"*30)
````

- 前端(Local)
    - 修出回應(網頁)
        - HTML: 內容
        - CSS: 排版
        - JavaScript: 網頁動態
- 後端(Server)
    - 收網址
    - 程式處理網址(Flask,Django,FastAPI:Python)
        - 查詢資料庫
        - 讀取檔案
    - 把回應(html/json...)送回去

- CSS語法: 找到html區塊並加上排版
    - 找所有html名字
        ```
        a {
            backgroupd-color: red
        }
        ```
    - 為了區分要的區塊,會在html區塊加上特殊屬性class(class = "分類一 分類二")
        ```
        .分類名字 {
            backgroupd-color: red
        }
        ```
    - 如果要獨一無二的值,使用id (id="獨一無二的")
        ```
        #id名字 {
            backgroupd-color: red
        }
        ```
